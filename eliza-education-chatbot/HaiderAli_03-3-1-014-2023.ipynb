#**Import libraries**

```python
import re
import random
```

#**Create rules for responses in a dictionary**

```python
botName = "TutorBot"

# Helpful tips database
study_tips = [
    "Try studying in 25-minute focused sessions with short breaks.",
    "Break big topics into small parts and learn one at a time.",
    "Teach the concept out loud — it improves understanding.",
    "Practice recalling information instead of rereading notes.",
    "Review material the same day you learn it."
]

motivation_tips = [
    "Start with just 5 minutes — getting started is the hardest part.",
    "Small progress every day leads to big success.",
    "Remove distractions and focus on one task only.",
    "Remember why you started — your goals matter."
]

responses = [

    # Greeting
    (r'(hi|hello|hey|salam)',
     "Hello! I'm your study assistant. What do you need help with?"),

    # Exams
    (r'(exam|test|quiz)',
     lambda: f"Exams can be stressful. {random.choice(study_tips)}"),

    # Study help
    (r'(study|learn|education)',
     lambda: f"Studying becomes easier with the right method. {random.choice(study_tips)}"),

    # Focus & distraction
    (r'(focus|distract|concentrate)',
     "Try studying in a quiet place and turn off phone notifications."),

    # Memory problems
    (r'(forget|memory|remember)',
     "Use active recall and spaced repetition to improve memory."),

    # Time management
    (r'(time|schedule|deadline)',
     "Make a simple daily plan and prioritize the most important task."),

    # Motivation / procrastination
    (r'(lazy|procrastinate|motivation)',
     lambda: random.choice(motivation_tips)),

    # Stress & anxiety
    (r'(stress|anxious|overwhelm|pressure)',
     "Take a short break, breathe slowly, and return with a clear mind."),

    # Coding / assignments
    (r'(assignment|project|homework|code|program)',
     "Break the task into small steps. If there's an error, share it and we can debug it together."),

    # Career / future
    (r'(career|job|internship)',
     "Building skills and projects will improve your career opportunities."),

    # Thanks
    (r'(thank)',
     "You're welcome! Keep working hard and stay consistent."),

    # Goodbye
    (r'(bye)',
     "Goodbye! Keep learning and improving every day."),

    # Default fallback
    (r'(.*)',
     "Tell me more about your study problem so I can help.")
]

```

#**Now creating a function to match patterns with specific keywords**

```python

def elizaResponse(userInput):
    for pattern, response in responses:
        if re.search(pattern, userInput.lower()):
            if callable(response):
                return response()
            else:
                return response
    return "I didn't understand that."

```
