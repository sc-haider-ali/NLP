#**Import libraries**

```python
# for pattern matching
import re
# for random responses
import random
```

#**Create rules for responses in a dictionary**

```python
patterns = {

    # GREETING
    r"\b(hi|hello|hey|salam|assalamualaikum)\b":
    [
        "Hello — I’m really glad you reached out. What has been challenging about your studies lately?",
        "Hi there. How have you been feeling about your learning and productivity recently?",
        "Hey! Tell me what’s been on your mind when it comes to studying or staying focused."
    ],

    # GRATITUDE
    r"\bthank\w*\b":
    [
        "You're very welcome. I'm glad it helped — keep moving forward one step at a time.",
        "Anytime. Supporting your learning journey is important.",
        "I’m happy to help. Remember, small improvements add up over time."
    ],

    # STUDY / LEARNING GENERAL
    r"\bstud\w*|\blearn\w*|\beducat\w*\b":
    [
        "Learning is a gradual and layered process. Which part of studying feels most difficult right now?",
        "Understanding develops step by step, not all at once. Give yourself time to absorb what you learn.",
        "Studying becomes more effective when you break complex topics into smaller, manageable pieces."
    ],

    # FAST LEARNING / EFFECTIVENESS
    r"\b(fast|quick\w*|speed)\b.*\b(stud\w*|learn\w*)\b|\b(stud\w*|learn\w*)\b.*\b(fast|quick\w*)\b":
    [
        "If learning feels slow, try working in short periods of deep focus followed by small breaks to keep your mind fresh.",
        "Many students absorb information faster when they study in focused intervals and pause briefly to recharge.",
        "Speed improves when focus improves — short, distraction-free sessions can make learning more efficient."
    ],

    # FOCUS & DISTRACTION
    r"\bfocus|concentrat\w*|distract\w*|attention\b":
    [
        "It can be frustrating to want to focus but feel pulled away. Small changes in your environment can make a big difference.",
        "Try committing to a short period of uninterrupted focus and allowing distractions to wait.",
        "Focus improves when your study space, mindset, and goals are clear and simple."
    ],

    # MEMORY & FORGETTING
    r"\bforget\w*|mem\w*|recall|retain\w*\b":
    [
        "Forgetting can feel discouraging, but it is a natural part of how memory strengthens.",
        "Reviewing material shortly after learning and recalling it without looking can improve retention.",
        "Understanding deepens through repetition — forgetting does not mean you failed to learn."
    ],

    # TIME MANAGEMENT & ROUTINE
    r"\btime|schedul\w*|plan\w*|routine|deadline\b":
    [
        "It sounds like you're trying to gain control over your time. Creating a simple daily structure can help reduce overwhelm.",
        "When everything feels urgent, choosing a few priorities for the day can restore clarity.",
        "A consistent routine, even if small, can create stability and steady progress."
    ],

    # SELF-DOUBT & IMPOSTER FEELINGS
    r"\b(useless|worthless|failure|dumb|not smart)\b|learned nothing|know nothing":
    [
        "That feeling can be very heavy. Sometimes learning something new makes us more aware of what we don’t yet know.",
        "Many students experience self-doubt even while improving — it doesn’t mean you aren’t learning.",
        "Progress in learning is often quiet and gradual. Your effort still matters, even when results feel invisible."
    ],

    # MOTIVATION & PROCRASTINATION
    r"\blazy|procrastinat\w*|motivat\w*|unmotivated\b":
    [
        "It can be hard to begin when motivation feels low. Starting with something very small can make the task less overwhelming.",
        "We often wait for motivation to appear, but beginning gently can create the momentum we need.",
        "Try committing to just a few minutes — once you start, continuing often feels easier."
    ],

    # STRESS & OVERWHELM
    r"\bstress\w*|overwhelm\w*|burnout|tired|pressure|anxious\b":
    [
        "It sounds like you're carrying a lot right now. Feeling overwhelmed is a sign that you may need rest and support.",
        "When stress builds, short pauses and slow breathing can help your body settle.",
        "Rest is not wasted time — it allows your mind to recover and return with clarity."
    ],

    # EXAMS & TEST ANXIETY
    r"\bexam\w*|test|quiz|paper|assessment\b":
    [
        "Exam pressure can feel intense, but steady preparation can gradually build confidence.",
        "Practicing questions and reviewing regularly can make exams feel more familiar and manageable.",
        "Anxiety often decreases when preparation becomes consistent and structured."
    ],

    # GOODBYE
    r"\b(bye|exit|quit|allah hafiz)\b":
    [
        "Take care of yourself and be patient with your progress.",
        "Goodbye for now — keep moving forward at your own pace.",
        "See you later. Remember, learning is a journey, not a race."
    ]
}
```

#**Now creating a function to match patterns with specific keywords**

```python
def chatbot_response(user_input):
    user_input = user_input.lower()

    for pattern, responses in patterns.items():
        if re.search(pattern, user_input):
            return random.choice(responses)

    return "I’m not sure I understand. Could you explain more about your study challenge?"
```

#**Now creating a loop to run the chatbot until you stop it.**

```python
print("Study Helper Bot 🤖")
print("Type 'exit' to exit.\n")

while True:
    user_input = input("You: ")

    response = chatbot_response(user_input)
    print("Bot:", response)

    if re.search(r"bye|exit|quit", user_input.lower()):
        break
```
